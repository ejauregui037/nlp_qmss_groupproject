```python
import os
import pdfplumber
import pandas as pd

# Function to extract text from a single PDF using pdfplumber
def extract_text_from_pdf(pdf_file):
    text = ''
    with pdfplumber.open(pdf_file) as pdf:
        # Extract text from each page
        for page in pdf.pages:
            text += page.extract_text()
    return text

# Function to process all PDFs in a directory
def process_pdfs_in_directory(directory_path, output_csv):
    # Check if the directory exists
    if not os.path.isdir(directory_path):
        print(f"Directory {directory_path} does not exist.")
        return

    # Initialize a list to hold the data
    pdf_data = []

    # Loop through all files in the directory
    for filename in os.listdir(directory_path):
        if filename.endswith('.pdf'):
            pdf_path = os.path.join(directory_path, filename)
            
            # Extract text from the PDF
            text = extract_text_from_pdf(pdf_path)
            
            # Append filename and text to the data list
            pdf_data.append([filename, text])

    # Create a DataFrame from the collected data
    df = pd.DataFrame(pdf_data, columns=["Filename", "Extracted Text"])


    return df 

```

```python

# Example usage
directory_path = "/Users/haleychen/Downloads/nlp_ai "  # Your PDF directory path
output_csv = "NLP_pdf.csv"  # Replace with your desired output CSV filename

df=process_pdfs_in_directory(directory_path, output_csv)
    
```

```python
import csv
import pandas as pd
import re

# Make a copy of the original DataFrame
cleaned_df = df.copy()

# Clean column names: remove special characters and strip whitespace
cleaned_df.columns = cleaned_df.columns.str.strip().str.replace(r'[^\w\s]', '', regex=True)

# Clean each string cell: strip whitespace and remove special characters
#also remove long space 
cleaned_df = cleaned_df.applymap(
    lambda x: re.sub(r'\s+', ' ', re.sub(r'[^\w\s]', '', str(x).strip())).strip() if isinstance(x, str) else x
)

# Preview the cleaned DataFrame
print(cleaned_df.head(20))

# Write the cleaned DataFrame to a CSV file
cleaned_df.to_csv(output_csv, index=False, encoding='utf-8')

print(f"CSV file '{output_csv}' has been generated successfully.")
```
